```python
class NeuralNetwork:
    def __init__(self, input_size, hidden_size, learning_rate=0.01):
        self.learning_rate = learning_rate

        self.w_hidden = np.random.randn(input_size, hidden_size) * learning_rate
        self.b_hidden = np.zeros((1, hidden_size))

        self.w_output = np.random.randn(hidden_size, 1) * learning_rate
        self.b_output = np.zeros((1, 1))

    def sigmoid(self, z):
        z = np.clip(z, -500, 500)
        return 1 / (1 + np.exp(-z))

```
